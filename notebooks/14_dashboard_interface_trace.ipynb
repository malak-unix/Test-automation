# Milestone 14 - Dashboard / Interface MVP

Milestone 14 adds a Flask dashboard that launches the existing full LangGraph workflow and displays the generated HTML report.

## Dashboard Role

The dashboard is a user interface layer only. It does not add new testing agents. It calls the existing workflow, summarizes the final State, lists recent runs, and embeds the generated HTML report.

## Why After Report Integration

The interface is useful only after the Report Agent can generate `final_results.json`, `report_result.json`, and `report_<run_id>.html` from a full integrated run.

## Interface Workflow

`User form -> run_workflow -> final_results/report_html -> displayed report`

Current main LangGraph workflow:

`START -> orchestrator -> repo_analyzer -> rag -> test_planner -> api_testing -> bug_analysis -> report -> END`

## Safety Note

The dashboard does not execute repository code. It only calls the existing workflow. It does not start Django, Docker, Selenium, Locust, MCP, or any LLM API by default.

Target repository: https://github.com/Vitaee/DjangoRestAPI

Target URL: http://localhost:8000
